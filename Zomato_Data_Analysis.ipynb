# ZOMATO DATA ANALYSIS PROJECT

Step 1 - Importing Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

Step 2 - Create the data frame

```python
df = pd.read_csv("Zomato data .csv")
print(df.head())
```

```python
df
```

# Convert the data type of column - rate

```python
def newRate(val):
    val = str(val).split("/")
    val = val[0]
    return float(val)
df['rate'] = df['rate'].apply(newRate)
print(df.head())
```

```python
df.info()
```

# Restaurant with majority of customer order from.

Type of restaurant

```python
df.head()
```

```python
sns.countplot(x=df['listed_in(type)'],palette="Set2")
plt.xlabel("Type of Restaurant")
```

Conclusion  - majority of the restaurant falls in dinning category.

# Total votes each restaurant received from customers.

```python
df.head()
```

```python
grouped_Data = df.groupby('listed_in(type)')['votes'].sum()
result = pd.DataFrame({'votes': grouped_Data})
plt.plot(result,c="red",marker = "o")
plt.xlabel("Type of Restaurant",c='magenta',size=20)
plt.ylabel("Votes",c="m",size=20)
```

Conclusion - dinning restaurant has received maximum votes.

# Majority of the ratings that restaurants received.

```python
df.head()
```

```python
plt.hist(df['rate'],bins = 6 , color = 'y')
plt.title("Rating distribution")
plt.show()
```

Conclusion -  the majority restaurants received ratings from 3.5 to 4

# Analysis of couples’ average spending per online order

```python
df.head()
```

```python
couple_data = df['approx_cost(for two people)']
sns.countplot(x=couple_data , palette = "muted")
```

Conclusion - the majority of couples prefer restaurants with an approximate cost of 300 rupees

# Highest-rated ordering mode

```python
df.head()
```

```python
plt.figure(figsize = (6,6))
custom_colors = ["skyblue","orange"]
sns.boxplot(data = df ,x = 'online_order' , y = 'rate' , palette = custom_colors)
```

Conclusion - offline order received lower rating in comparison to online order.

# Restaurant type with most offline orders

```python
df.head(20)
```

```python
pivot_table = df.pivot_table(
    index = 'listed_in(type)' , 
    columns = 'online_order' , 
    aggfunc = 'size',
    fill_value = 0
)
sns.heatmap(pivot_table, annot = True , cmap = 'Spectral',fmt = 'd')
plt.title("Heatmap")
plt.xlabel("Online order")
plt.ylabel("Listed in (Type)")
plt.show()
```

Conclusion - Dining resturants primarily accept offline orders , whereas cafes primarily receive online orders. This suggests that clients prefers to place orders in person at restaurants , but prefer online ordering at cafes.
